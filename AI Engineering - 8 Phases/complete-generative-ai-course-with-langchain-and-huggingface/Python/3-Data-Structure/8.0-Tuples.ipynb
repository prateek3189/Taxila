# Tuples

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(lst)

tpl=tuple()
print(tpl)
```

```python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)
```

```python
mixed_tuple = ("Start", 1, "Hello", 3.4, "end")
print(mixed_tuple)
```

```python
numbers = (1, 2, 3, 4, 5)
print(numbers)

print(numbers[0])
```

```python
concatinate = numbers + mixed_tuple
print(concatinate)
```

```python
mixed_tuple * 3
```

```python
print(numbers.count(3))
print(numbers.index(3))
```

```python
packed_tuple = 1, 2, 3, 4, 5
print(packed_tuple)
```

```python
a, b, c, d, e = packed_tuple
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
first, *middle, second_last, last = packed_tuple
print(first)
print(middle)
print(second_last)
print(last)
```
