```python
fruits=["Apple", "Banana", "Cherry", "Mango", "Orange", "Pineapple", "Strawberry", "Watermelon"]
print(fruits[0])
print(fruits[-1])
print(fruits[1:])
print(fruits[:2])
print(fruits[1:3])

fruits[0] = "Grapes"
print(fruits)

fruits.append("Kiwi")
print(fruits)

fruits.insert(1, "Papaya")
print(fruits)

fruits.remove("Mango")
print(fruits)

fruits.pop()
print(fruits)

index = fruits.index("Banana")
print(index)

count = fruits.count("Banana")
print(count)

fruits.sort()
print(fruits)

fruits.reverse()
print(fruits)
print(len(fruits))

fruits.clear()
print(fruits)
```

```python
## List Comprehension
[x**2 for x in range(10)]
```

```python
square = [numb**2 for numb in range(10)]
```

```python
lst1 = [1, 2, 3]
lst2 = ['a', 'b', 'c']
combined = [(x, y) for x in lst1 for y in lst2]
print(combined)
```
