```python
## Basic Syntax Rules In Python
## Case Sensitivity

name = "John"
Name = "Doe"
print(name)
print(Name)
```

```python
## Indentation
age = 37
if age > 18:
    print("You are an adult.")
```

```python
## This is a single line comment
print("Hello, World!")  # This is an inline comment
```

```python
## Line Continuation

total = 1 + 2 + 3 + \
        4 + 5
print(total)
```

```python
## Multiple statement in a single line
x=5;y=10;z=x+y
print(z)
```

```python
## Understanding Semantics in Python
## Variables Assignment
x = 5
y = "Hello, World!"
print(x)
print(y)
print(type(x))
print(type(y))
```

```python
# Code Examples of indentation
if True:
    print("Correct Indentation")
    if False:
        print("This will not be printed") 
    print("Still Correct Indentation")
print("Outside If Block")  # This will raise an IndentationError
```
