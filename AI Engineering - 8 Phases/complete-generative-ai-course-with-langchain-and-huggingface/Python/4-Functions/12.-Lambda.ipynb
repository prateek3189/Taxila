# Lambda

```python
addition = lambda a, b: a + b
type(addition)  # <class 'function'>

print(addition(5, 3))  # Output: 8
```

```python
def even(num):
    return num % 2 == 0

# Using lambda
even_lambda = lambda num: num % 2 == 0
print(even_lambda(4))  # Output: True
```

```python
numbers = [1, 2, 3, 4, 5, 6]
list(map(lambda x: x * 2, numbers))  # Output: [2, 4, 6, 8, 10, 12]
```
