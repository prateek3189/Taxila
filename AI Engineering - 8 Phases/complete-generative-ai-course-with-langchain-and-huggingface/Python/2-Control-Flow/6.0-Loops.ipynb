## Loops

```python
## For Loop
for i in range(5):
    print(i)
```

```python
for i in range(1, 5):
    print(i)
```

```python
for i in range(1, 20, 2):
    print(i)
```

```python
str = "Prateek Magarde"

for i in str:
    print(i)
```

```python
## While Loop
i = 1
while i < 6:
    print(i)
    i += 1
```

```python
for i in range(5):
  if i == 3:
    break
  print(i)
```

```python
for i in range(10):
  if i%2 == 0:
    continue
  print(i)
```

```python
for i in range(10):
  if i == 3:
    pass
  print(i)
```

```python
for i in range(5):
  for j in range(3):
    print(f"i: {i} and j: {j}")
```
