### Exception Handling

```python
try:
  a = b
except:
  print("The Variable is Undefined")
```

```python
try:
  a = b
except NameError as ex:
  print(ex)
```

```python
try:
  res = 1/0
except Exception as ex:
  print(ex)
```

```python
try:
  num = int(input("Enter a number"))
  res = 10/num
except Exception as ex:
  print(ex)
else:
  print(res)
```

```python
try:
  num = int(input("Enter a number"))
  res = 10/num
except Exception as ex:
  print(ex)
else:
  print(res)
finally:
  print("Completed!!")
```

```python
# File Handling and Exception Handling
try:
  file = open("example.txt", "r")
  content = file.read();
  print(content)
except FileNotFoundError as ex:
  print(ex)
finally:
  if "file" in locals() or not file.closed():
    file.close()
    print("File Closed")
```
